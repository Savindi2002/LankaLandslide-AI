```python
import os
import pandas as pd
import numpy as np

print("LankaLandslide AI")
print("Cleaning and merging data...")
```

```python
BASE_DIR = os.path.abspath("..")

LANDSLIDE_FILE = os.path.join(
    BASE_DIR,
    "data",
    "processed",
    "landslides_sri_lanka.csv"
)

WEATHER_FILE = os.path.join(
    BASE_DIR,
    "data",
    "raw",
    "weather",
    "sri_lanka_complete_daily_weather_1998_2025.csv"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "data",
    "processed"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

MASTER_FILE = os.path.join(
    OUTPUT_DIR,
    "landslide_weather_master.csv"
)

print("Landslide file:")
print(LANDSLIDE_FILE)

print("\nWeather file:")
print(WEATHER_FILE)

print("\nMaster dataset:")
print(MASTER_FILE)
```

Load landslide data

```python
landslides = pd.read_csv(
    LANDSLIDE_FILE,
    parse_dates=["date"]
)

print("Landslide dataset loaded.")
print("Rows:", len(landslides))
print("Columns:", len(landslides.columns))

landslides.head()
```

Load weather data

```python
weather = pd.read_csv(
    WEATHER_FILE,
    parse_dates=["date"]
)

print("Weather dataset loaded.")
print("Rows:", len(weather))
print("Columns:", len(weather.columns))

weather.head()
```

Inspect both datasets

```python
print("LANDSLIDE DATA")
print("================")
print(landslides.info())

print("\n\nWEATHER DATA")
print("================")
print(weather.info())
```

Standardize district names

```python
landslides["district"] = (
    landslides["district"]
    .astype(str)
    .str.strip()
)

weather["district"] = (
    weather["district"]
    .astype(str)
    .str.strip()
)

print("Landslide districts:")
print(sorted(landslides["district"].unique()))

print("\nWeather districts:")
print(sorted(weather["district"].unique()))
```

Check district matching

```python
landslide_districts = set(
    landslides["district"].dropna().unique()
)

weather_districts = set(
    weather["district"].dropna().unique()
)

print("Districts in both datasets:")
print(sorted(
    landslide_districts.intersection(weather_districts)
))

print("\nLandslide districts missing from weather:")
print(sorted(
    landslide_districts - weather_districts
))

print("\nWeather districts without landslide records:")
print(sorted(
    weather_districts - landslide_districts
))
```

Make sure dates are correct

```python
landslides["date"] = pd.to_datetime(
    landslides["date"],
    errors="coerce"
)

weather["date"] = pd.to_datetime(
    weather["date"],
    errors="coerce"
)

print("Landslide date range:")
print(
    landslides["date"].min(),
    "→",
    landslides["date"].max()
)

print("\nWeather date range:")
print(
    weather["date"].min(),
    "→",
    weather["date"].max()
)
```

Remove unusable landslide records

```python
before = len(landslides)

landslides = landslides.dropna(
    subset=["date", "district"]
).copy()

after = len(landslides)

print("Records before:", before)
print("Records after:", after)
print("Removed:", before - after)
```

reate daily landslide indicator

```python
landslide_daily = (
    landslides
    .groupby(["district", "date"])
    .size()
    .reset_index(name="landslide_events")
)

landslide_daily["landslide_recorded"] = 1

print("Unique district-date landslide records:")
print(len(landslide_daily))

landslide_daily.head()
```

Check multiple events

```python
print(
    "District-date combinations with multiple landslide records:"
)

print(
    (landslide_daily["landslide_events"] > 1).sum()
)
```

Restrict training period

```python
print(
    landslides["date"]
    .dt.year
    .value_counts()
    .sort_index()
)
```

```python
TRAIN_START = landslides["date"].min()
TRAIN_END = landslides["date"].max()

print("Training period:")
print(TRAIN_START, "→", TRAIN_END)
```

Filter weather to the overlapping period

```python
weather_train = weather[
    (weather["date"] >= TRAIN_START) &
    (weather["date"] <= TRAIN_END)
].copy()

print("Weather rows for training period:")
print(len(weather_train))

print(
    "Date range:",
    weather_train["date"].min(),
    "→",
    weather_train["date"].max()
)
```

Merge landslide records with weather

```python
master = weather_train.merge(
    landslide_daily[
        [
            "district",
            "date",
            "landslide_events",
            "landslide_recorded"
        ]
    ],
    on=["district", "date"],
    how="left"
)
```

Create the target variable

```python
master["landslide_events"] = (
    master["landslide_events"]
    .fillna(0)
    .astype(int)
)

master["landslide_recorded"] = (
    master["landslide_recorded"]
    .fillna(0)
    .astype(int)
)

print(
    master["landslide_recorded"]
    .value_counts()
)
```

Check the master dataset

```python
print("========================================")
print("MASTER DATASET")
print("========================================")

print("Rows:", len(master))
print("Columns:", len(master.columns))

print("\nDate range:")
print(
    master["date"].min(),
    "→",
    master["date"].max()
)

print("\nDistricts:")
print(master["district"].nunique())

print("\nTarget distribution:")
print(
    master["landslide_recorded"]
    .value_counts()
)

print("\nMissing values:")
print(
    master.isna().sum()
)
```

Check duplicates

```python
duplicates = master.duplicated(
    subset=["district", "date"]
).sum()

print(
    "Duplicate district-date rows:",
    duplicates
)
```

Missing values — inspect first

```python
print("LANDSLIDE MISSING VALUES")
print("========================")
print(landslides.isna().sum())

print("\nWEATHER MISSING VALUES")
print("======================")
print(weather.isna().sum())
```

```python
# ============================================================
# COMPLETE MISSING VALUE REPORT
# ============================================================

landslide_missing = pd.DataFrame({
    "missing_count": landslides.isna().sum(),
    "missing_percent": (landslides.isna().mean() * 100).round(2)
})

landslide_missing = landslide_missing[
    landslide_missing["missing_count"] > 0
].sort_values("missing_count", ascending=False)

print("LANDSLIDE MISSING VALUES")
print("=" * 60)
print(landslide_missing.to_string())
```

Check whether latitude/longitude are actually useful

```python
# ============================================================
# CHECK INVALID COORDINATES
# ============================================================

invalid_coordinates = landslides[
    (landslides["latitude"] == 0) |
    (landslides["longitude"] == 0)
]

print("Records with 0 latitude/longitude:")
print(len(invalid_coordinates))

print("\nPercentage:")
print(round(len(invalid_coordinates) / len(landslides) * 100, 2), "%")
```

```python
# ============================================================
# WEATHER MISSING VALUES
# ============================================================

weather_missing = pd.DataFrame({
    "missing_count": weather.isna().sum(),
    "missing_percent": (weather.isna().mean() * 100).round(2)
})

weather_missing = weather_missing[
    weather_missing["missing_count"] > 0
].sort_values("missing_count", ascending=False)

print("WEATHER MISSING VALUES")
print("=" * 60)
print(weather_missing.to_string())
```

```python
def missing_report(df, name):

    report = pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_percent": (
            df.isna().mean() * 100
        ).round(2)
    })

    print(f"\n{name}")
    print("=" * 50)
    print(report.sort_values(
        "missing_percent",
        ascending=False
    ))

missing_report(landslides, "LANDSLIDE DATA")
missing_report(weather, "WEATHER DATA")
```

Weather missing values

```python
weather_missing = weather[
    weather.isna().any(axis=1)
]

print("Weather rows with missing values:")
print(len(weather_missing))

weather_missing.head()
```

Check invalid numerical values

```python
print("Rainfall below zero:")
print((weather["rainfall_mm"] < 0).sum())

print("Temperature impossible values:")
print(
    (
        (weather["temp_min"] < -20) |
        (weather["temp_min"] > 60)
    ).sum()
)

print("Wind below zero:")
print((weather["wind_max"] < 0).sum())
```

Detect rainfall outliers using IQR

```python
Q1 = weather["rainfall_mm"].quantile(0.25)
Q3 = weather["rainfall_mm"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

rainfall_outliers = weather[
    (weather["rainfall_mm"] < lower_bound) |
    (weather["rainfall_mm"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("\nLower bound:", lower_bound)
print("Upper bound:", upper_bound)

print("\nPotential rainfall outliers:")
print(len(rainfall_outliers))
```

Inspect the largest rainfall values

```python
weather[
    [
        "date",
        "district",
        "rainfall_mm"
    ]
].sort_values(
    "rainfall_mm",
    ascending=False
).head(20)
```

Check weather outliers

```python
def iqr_report(df, column):

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(f"\n{column}")
    print("-" * 40)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower:", lower)
    print("Upper:", upper)
    print("Potential outliers:", len(outliers))


iqr_report(weather, "rainfall_mm")
iqr_report(weather, "temp_max")
iqr_report(weather, "temp_min")
iqr_report(weather, "wind_max")
iqr_report(weather, "wind_gust_max")
```

Create the target variable

```python
master["landslide_events"] = (
    master["landslide_events"]
    .fillna(0)
    .astype(int)
)

master["landslide_recorded"] = (
    master["landslide_recorded"]
    .fillna(0)
    .astype(int)
)

print(
    master["landslide_recorded"]
    .value_counts()
)
```

Check the master dataset

```python
print("========================================")
print("MASTER DATASET")
print("========================================")

print("Rows:", len(master))
print("Columns:", len(master.columns))

print("\nDate range:")
print(
    master["date"].min(),
    "→",
    master["date"].max()
)

print("\nDistricts:")
print(master["district"].nunique())

print("\nTarget distribution:")
print(
    master["landslide_recorded"]
    .value_counts()
)

print("\nMissing values:")
print(
    master.isna().sum()
)
```

Save the master dataset

```python
master = master.sort_values(
    ["district", "date"]
).reset_index(drop=True)

master.to_csv(
    MASTER_FILE,
    index=False
)

print("========================================")
print("MASTER DATASET SAVED")
print("========================================")

print(MASTER_FILE)
print("Rows:", len(master))
print("Columns:", len(master.columns))
```
