```python
import os
import time
import requests
import pandas as pd
from datetime import datetime

print("LankaLandslide AI")
print("Weather data collection started.")
```

```python
BASE_DIR = os.path.abspath("..")

WEATHER_RAW_DIR = os.path.join(
    BASE_DIR,
    "data",
    "raw",
    "weather"
)

os.makedirs(WEATHER_RAW_DIR, exist_ok=True)

print("Project directory:")
print(BASE_DIR)

print("\nWeather data directory:")
print(WEATHER_RAW_DIR)
```

```python
districts = {
    "Colombo": (6.9271, 79.8612),
    "Gampaha": (7.0840, 80.0098),
    "Kalutara": (6.5854, 79.9607),
    "Kandy": (7.2906, 80.6337),
    "Matale": (7.4675, 80.6234),
    "Nuwara Eliya": (6.9497, 80.7891),
    "Kegalle": (7.2513, 80.3464),
    "Ratnapura": (6.6828, 80.3992),
    "Galle": (6.0329, 80.2168),
    "Matara": (5.9549, 80.5550),
    "Hambantota": (6.1429, 81.1212),
    "Badulla": (6.9934, 81.0550),
    "Monaragala": (6.8728, 81.3507),
    "Kurunegala": (7.4863, 80.3647),
    "Puttalam": (8.0362, 79.8283),
    "Anuradhapura": (8.3114, 80.4037),
    "Polonnaruwa": (7.9403, 81.0188),
    "Jaffna": (9.6615, 80.0255),
    "Kilinochchi": (9.3803, 80.3770),
    "Mannar": (8.9810, 79.9044),
    "Vavuniya": (8.7514, 80.4971),
    "Trincomalee": (8.5874, 81.2152),
    "Batticaloa": (7.7170, 81.7000),
    "Ampara": (7.2917, 81.6720)
}

print("Number of districts:", len(districts))

for district, coordinates in districts.items():
    print(district, ":", coordinates)
```

```python
START_DATE = "1998-01-01"
END_DATE = "2025-12-31"

API_URL = "https://archive-api.open-meteo.com/v1/archive"

OUTPUT_FILE = os.path.join(
    WEATHER_RAW_DIR,
    "sri_lanka_complete_daily_weather_1998_2025.csv"
)

print("Start date:", START_DATE)
print("End date:", END_DATE)
print("Output file:", OUTPUT_FILE)
```

```python
DAILY_VARIABLES = [
    "precipitation_sum",
    "rain_sum",
    "precipitation_hours",
    "temperature_2m_max",
    "temperature_2m_min",
    "temperature_2m_mean",
    "wind_speed_10m_max",
    "wind_gusts_10m_max"
]

print("Weather variables:")
for variable in DAILY_VARIABLES:
    print("-", variable)
```

```python
test_district = "Kandy"

latitude, longitude = districts[test_district]

params = {
    "latitude": latitude,
    "longitude": longitude,
    "start_date": "2024-01-01",
    "end_date": "2024-01-10",
    "daily": ",".join(DAILY_VARIABLES),
    "timezone": "Asia/Colombo",
    "models": "era5"
}

response = requests.get(
    API_URL,
    params=params,
    timeout=120
)

print("Status code:", response.status_code)

if response.status_code == 200:
    data = response.json()

    print("API connection successful!")
    print("Latitude:", data["latitude"])
    print("Longitude:", data["longitude"])

    print("\nAvailable daily variables:")
    print(data["daily"].keys())

else:
    print("API request failed.")
    print(response.text[:500])
```

```python
daily = data["daily"]

test_df = pd.DataFrame(daily)

test_df["date"] = pd.to_datetime(test_df["time"])

test_df = test_df.rename(columns={
    "precipitation_sum": "rainfall_mm",
    "rain_sum": "rain_mm",
    "precipitation_hours": "rain_hours",
    "temperature_2m_max": "temp_max",
    "temperature_2m_min": "temp_min",
    "temperature_2m_mean": "temp_mean",
    "wind_speed_10m_max": "wind_max",
    "wind_gusts_10m_max": "wind_gust_max"
})

test_df = test_df.drop(columns=["time"])

test_df["district"] = test_district
test_df["latitude"] = latitude
test_df["longitude"] = longitude

test_df = test_df[
    [
        "date",
        "rainfall_mm",
        "rain_mm",
        "rain_hours",
        "temp_max",
        "temp_min",
        "temp_mean",
        "wind_max",
        "wind_gust_max",
        "district",
        "latitude",
        "longitude"
    ]
]

print(test_df)
```

```python
import calendar

TEMP_FILE = os.path.join(
    WEATHER_RAW_DIR,
    "weather_download_progress.csv"
)

# Load existing progress if available
if os.path.exists(TEMP_FILE):
    weather_all = pd.read_csv(
        TEMP_FILE,
        parse_dates=["date"]
    )

    print("Existing progress loaded.")
    print("Rows:", len(weather_all))

else:
    weather_all = pd.DataFrame(
        columns=[
            "date",
            "rainfall_mm",
            "rain_mm",
            "rain_hours",
            "temp_max",
            "temp_min",
            "temp_mean",
            "wind_max",
            "wind_gust_max",
            "district",
            "latitude",
            "longitude"
        ]
    )

print("Starting resumable weather download...")
```

```python
for district, (latitude, longitude) in districts.items():

    print("\n" + "=" * 60)
    print("DISTRICT:", district)
    print("=" * 60)

    for year in range(1998, 2026):

        # Check whether this district-year already exists
        already_done = (
            (weather_all["district"] == district) &
            (pd.to_datetime(weather_all["date"]).dt.year == year)
        ).any()

        if already_done:
            print(f"{year}: already completed → skipping")
            continue

        start_date = f"{year}-01-01"
        end_date = f"{year}-12-31"

        params = {
            "latitude": latitude,
            "longitude": longitude,
            "start_date": start_date,
            "end_date": end_date,
            "daily": ",".join(DAILY_VARIABLES),
            "timezone": "Asia/Colombo",
            "models": "era5"
        }

        success = False

        for attempt in range(1, 6):

            try:

                response = requests.get(
                    API_URL,
                    params=params,
                    timeout=120
                )

                print(
                    f"{district} | {year} | "
                    f"Attempt {attempt} | "
                    f"Status {response.status_code}"
                )

                if response.status_code == 200:

                    api_data = response.json()
                    daily = api_data["daily"]

                    year_df = pd.DataFrame(daily)

                    year_df["date"] = pd.to_datetime(
                        year_df["time"]
                    )

                    year_df = year_df.rename(columns={
                        "precipitation_sum": "rainfall_mm",
                        "rain_sum": "rain_mm",
                        "precipitation_hours": "rain_hours",
                        "temperature_2m_max": "temp_max",
                        "temperature_2m_min": "temp_min",
                        "temperature_2m_mean": "temp_mean",
                        "wind_speed_10m_max": "wind_max",
                        "wind_gusts_10m_max": "wind_gust_max"
                    })

                    year_df = year_df.drop(
                        columns=["time"]
                    )

                    year_df["district"] = district
                    year_df["latitude"] = latitude
                    year_df["longitude"] = longitude

                    year_df = year_df[
                        [
                            "date",
                            "rainfall_mm",
                            "rain_mm",
                            "rain_hours",
                            "temp_max",
                            "temp_min",
                            "temp_mean",
                            "wind_max",
                            "wind_gust_max",
                            "district",
                            "latitude",
                            "longitude"
                        ]
                    ]

                    weather_all = pd.concat(
                        [weather_all, year_df],
                        ignore_index=True
                    )

                    # Remove accidental duplicates
                    weather_all = weather_all.drop_duplicates(
                        subset=["district", "date"]
                    )

                    weather_all = weather_all.sort_values(
                        ["district", "date"]
                    ).reset_index(drop=True)

                    # SAVE AFTER EVERY YEAR
                    weather_all.to_csv(
                        TEMP_FILE,
                        index=False
                    )

                    print(
                        f"✓ {district} {year} saved "
                        f"({len(year_df)} days)"
                    )

                    success = True
                    break

                elif response.status_code == 429:

                    wait_time = 20 * attempt

                    print(
                        f"Rate limited. "
                        f"Waiting {wait_time} seconds..."
                    )

                    time.sleep(wait_time)

                else:

                    print(
                        "API error:",
                        response.text[:300]
                    )

                    time.sleep(5)

            except Exception as e:

                print("Error:", e)

                time.sleep(10)

        if not success:

            print(
                f"⚠ Could not download "
                f"{district} {year}"
            )

        # Small delay between requests
        time.sleep(2)

print("\n======================================")
print("DOWNLOAD PROCESS FINISHED")
print("======================================")

print("Total rows:", len(weather_all))
print("Districts:", weather_all["district"].nunique())
```

```python
final_weather_file = os.path.join(
    WEATHER_RAW_DIR,
    "sri_lanka_complete_daily_weather_1998_2025.csv"
)

weather_all = weather_all.drop_duplicates(
    subset=["district", "date"]
)

weather_all = weather_all.sort_values(
    ["district", "date"]
).reset_index(drop=True)

weather_all.to_csv(
    final_weather_file,
    index=False
)

print("Final weather dataset saved:")
print(final_weather_file)
```
